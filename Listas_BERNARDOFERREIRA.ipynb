```python
n = int(input("Quantas pessoas serão informadas? "))
mais_velho_nome = ""
mais_velho_idade = -1

for i in range(n):
    nome = input(f"Digite o nome da {i+1}ª pessoa: ")
    idade = int(input(f"Digite a idade de {nome}: "))
    if idade > mais_velho_idade:
        mais_velho_idade = idade
        mais_velho_nome = nome

print("Pessoa mais velha:", mais_velho_nome, "-", mais_velho_idade, "anos")
```

```python
n = int(input("Quantas pessoas serão informadas? "))
somas = {"M": 0, "F": 0}
conts = {"M": 0, "F": 0}

for i in range(n):
    sexo = input(f"Digite o sexo da {i+1}ª pessoa (M/F): ").upper()
    idade = int(input("Digite a idade: "))
    if sexo in somas:
        somas[sexo] += idade
        conts[sexo] += 1

for s in ["M", "F"]:
    if conts[s] > 0:
        print(f"Média de idade do sexo {s}: {somas[s] / conts[s]}")
```
